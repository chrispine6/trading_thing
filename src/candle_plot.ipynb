```python
import pandas as pd
import plotly.graph_objects as go
```

```python
df = pd.read_csv('../data/aapl.us.csv')
df['Date'] = pd.to_datetime(df['Date'])
df = df.iloc[:100]
print(df.head())
```

```python
# Create figure
fig = go.Figure(data=[go.Candlestick(
    x=df['Date'],
    open=df['Open'],
    high=df['High'],
    low=df['Low'],
    close=df['Close'],
    hovertext=[
        f"Open: {o}<br>High: {h}<br>Low: {l}<br>Close: {c}<br>Volume: {v}"
        for o, h, l, c, v in zip(df['Open'], df['High'], df['Low'], df['Close'], df['Volume'])
    ],
    hoverinfo="text"
)])

# Customize layout
fig.update_layout(
    title='AAPL Candlestick Chart (Interactive)',
    xaxis_title='Date',
    yaxis_title='Price (USD)',
    xaxis_rangeslider_visible=False,  # Optional: hide range slider
    template='plotly_dark',           # Optional: dark theme
)
```

```python
# Show figure
fig.show()
```
