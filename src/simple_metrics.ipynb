# simple metrics
## candle body strength
what does the body strength tell us tho??

```python
def calc_body_strength(open_val, close):
    body_strength = abs(close - open_val)
    return body_strength

def calc_body_strength_pct(open_val, close):
    body_strength_pct = abs(close - open_val) / open_val * 100
    return body_strength_pct
```

## win/loss ratio (directional bias)

in a 20 candle window, 50 green candles = 15% bull ratio. above 60-70% suggests directional dominance

```python
def win_loss_ratio(n_green, n_red, n_total):
    bull_ration = n_green / n_total * 100
    bear_ration = n_red / n_total * 100
    if n_green > n_red:
        directional_bias = 'bullish'
        return bull_ration, directional_bias
    elif n_red > n_green:
        directional_bias = 'bearish'
        return bear_ration, directional_bias
    else:
        directional_bias = 'neutral'
        return 50, directional_bias
```

## higher highs and higher lows (trend identification)
a series of higher highs and higher lows indicates an uptrend, while lower highs and lower lows indicate a downtrend

```python
def trend_identifier(n_consecutive_higher_highs, n_consecutive_higher_lows, n_consecutive_lower_highs, n_consecutive_lower_lows):
    if n_consecutive_higher_highs >= 3 and n_consecutive_higher_lows >= 3:
        trend = 'uptrend'
    elif n_consecutive_lower_highs >= 3 and n_consecutive_lower_lows >= 3:
        trend = 'downtrend'
    else:
        trend = 'sideways'
    return trend
```

## wick analysis (rejection strength)
if upper wick is >50% of total range, strong rejection at top. long lower wick in uptrend = buyers defending, bullish

```python
def rejection_strength(open_val, close, high, low):
    upper_wick = high - max(open_val, close)
    lower_wick = min(open_val, close) - low
    rejection_ratio = upper_wick / (high - low)
    return upper_wick, lower_wick, rejection_ratio
```

## momentum (rate of change)
is momentum increasing or decreasing? compare recent 5 day momentum to previous 5 day momentum

```python
close = []
def momentum(n):
    price_velocity = (close[n] - close[n-1]) / close[n-1]
    return price_velocity
```

## pull back magnitude
healthy pullbacks in uptrend = 10-40%
deep pullbacks (>50%): trend weakness

```python
def pullback(recent_high, pullback_low):
    pullback_percent = (recent_high - pullback_low) / recent_high * 100
    return pullback_percent
```
