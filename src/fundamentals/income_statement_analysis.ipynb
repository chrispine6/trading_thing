```python
import certifi
import json
import ssl
from urllib.request import urlopen
```

```python
def get_jsonparsed_data(url):
    context = ssl.create_default_context(cafile=certifi.where())
    response = urlopen(url, context=context)
    data = response.read().decode("utf-8")
    return json.loads(data)
```

```python
class OperatingMetrics():
    """
        operating income before tax (ebit)
        operating income = gross profit - operating expenses
        - this metric shows overall profits excluding tax affects
        - basic metric for comparing companies with different capital structure
    """
    def __init__(self, gross_profit, operating_expense, revenue):
        self.gross_profit = gross_profit
        self.operating_expenses = operating_expense
        self.revenue = revenue

    def operating_income(self):
        return self.gross_profit - self.operating_expenses

    def operating_margin(self):
        oi = self.operating_income()
        try:
            return (oi/self.revenue) * 100
        except ZeroDivisionError:
            return float('nan')
```
